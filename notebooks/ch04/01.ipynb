현재 폴더를 확인한다.

현재 폴더 안에 data 폴더가 있는지 확인한다.

만약 data 폴더가 있다면

    현재 폴더를 프로젝트 루트로 결정한다.

    탐색을 종료한다.

만약 data 폴더가 없다면

    부모 폴더로 한 단계 올라간다.

다시 data 폴더가 있는지 확인한다.

이 과정을 반복한다.

더 이상 위로 올라갈 수 없는데도

데이터 폴더를 찾지 못했다면

오류를 발생시킨다.

```python
from pathlib import Path
# 1. 현재 폴더를 가져온다.
current = Path.cwd().resolve()
print("현재 위치:", current)
# 2. data 폴더를 찾을 때까지 반복한다.
while True:
    # 현재 위치에 data 폴더가 있는지 확인한다.
    if (current / "data").is_dir():
        project_root = current
        break
    # 더 이상 부모 폴더로 올라갈 수 없다면 오류를 발생시킨다.
    if current.parent == current:
        raise FileNotFoundError(
            "data 폴더가 있는 프로젝트 루트를 찾을 수 없습니다."
        )
    # data 폴더가 없다면 부모 폴더로 한 단계 올라간다.
    current = current.parent
# 3. 결과를 확인한다.
print("프로젝트 루트:", project_root)
```

```python
def get_project_root(start_path=None):
    """data 폴더가 있는 프로젝트 루트를 찾는다."""
    current = Path(
        start_path or Path.cwd()
    ).resolve()
    while True:
        if (current / "data").is_dir():
            return current
        if current.parent == current:
            raise FileNotFoundError(
                "data 폴더가 있는 프로젝트 루트를 찾을 수 없습니다."
            )
        current = current.parent
def get_data_dir():
    return get_project_root() / "data" / "raw"

def get_report_dir():
    return get_project_root() / "reports"
```

```python
PROJECT_ROOT = get_project_root()
DATA_DIR = get_data_dir()
REPORT_DIR = get_report_dir()
print("프로젝트 루트:", PROJECT_ROOT)
print("데이터 폴더:", DATA_DIR)
print("결과 폴더:", REPORT_DIR)
```
