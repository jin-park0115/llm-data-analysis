## 실습 3. 분석에 필요한 컬럼만 선택하기


원본 데이터의 모든 컬럼이 이후 분석에 필요한 것은 아닙니다.

- 순위, 상품명, 판매가, 인물, 출판사, 분야 등의 정보를 확인할 수 있습니다.

## 실습 4. 컬럼 이름을 사용하기 쉽게 정리하기
 
원본 컬럼 이름은 사람이 읽기에는 좋지만 코드에서 반복해서 사용하기에는 조금 길 수 있습니다.

- `판매상품 ID` → `판매상품ID`
- `인물` → `저자`
- `발행(출시)일자` → `발행일`
- `columns`를 통해 변경된 컬럼명을 정상적으로 확인했습니다.

## 실습 7. 발행일을 날짜형으로 바꾸기

원본 데이터의 발행일은 다음처럼 보일 수 있습니다.

```python
# 날짜형으로 변환
df_books['발행일'] = pd.to_datetime(
    df_books['발행일'],
    format='%Y%m%d'
)

# 자료형 확인
print(df_books['발행일'].dtype)

# 앞의 5개 값 확인
print(df_books['발행일'].head())
```

### 실행 결과

- `dtype`이 **datetime64[us]**로 확인되어 날짜형 변환이 정상적으로 완료되었습니다.
- 앞의 5개 날짜도 정상적으로 확인했습니다.

## 실습 8. 문자열 앞뒤 공백 정리하기
 
상품명, 저자, 출판사, 분야는 문자열 데이터입니다.

문자열 앞뒤에 불필요한 공백이 있으면 같은 값도 서로 다른 데이터처럼 처리될 수 있습니다.

예를 들어 다음 두 값은 사람이 보기에는 같지만 문자열로는 다릅니다.

```python
# 공백을 제거할 컬럼
columns = ['상품명', '저자', '출판사', '분야']

# 각 컬럼의 앞뒤 공백 제거
for col in columns:
    df_books[col] = df_books[col].str.strip()

# 각 컬럼의 앞 5개 값 확인
for col in columns:
    print(f'\n{col}')
    print(df_books[col].head())
```

### 실행 결과

- `상품명`, `저자`, `출판사`, `분야` 컬럼의 앞뒤 공백을 제거했습니다.
- 각 컬럼의 상위 5개 데이터를 확인하여 정상적으로 처리된 것을 확인했습니다.

## 실습 9. 중복 데이터 확인하기

데이터에서 중복을 발견했다고 해서 바로 삭제하면 안 됩니다.

먼저 무엇을 기준으로 중복인지 생각해야 합니다.

예를 들어 같은 상품명이 여러 번 등장할 수 있습니다.

하지만 같은 제목이라도 다른 판본, 다른 상품, 개정판 등이 있을 수 있습니다.

따라서 이번 데이터에서는 상품명만 보고 중복 행을 바로 삭제하지 않습니다.

우선 상품을 구분하는 판매상품ID를 확인합니다.

```python
# 1. 전체 행이 완전히 같은 중복 개수
print(df_books.duplicated().sum())

# 2. 판매상품ID가 중복된 행 개수
print(df_books.duplicated(subset='판매상품ID').sum())

# 3. 상품명이 중복된 행 개수
print(df_books.duplicated(subset='상품명').sum())
```

도서 데이터(df_books)에서 중복 여부를 확인한 결과는 다음과 같습니다.
- 전체 컬럼 값이 완전히 동일한 행의 중복 개수: 0개
- 판매상품ID가 중복된 행의 개수: 0개
- 상품명이 중복된 행의 개수: 0개
따라서 현재 데이터에는 완전히 동일한 중복 행이 없으며, 각 도서의 판매상품ID와 상품명도 중복되지 않는 것으로 확인되었습니다.
