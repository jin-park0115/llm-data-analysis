## 실습 3. Dictionary로 단어 빈도 원리 이해하기

빈도 분석의 핵심은 매우 단순합니다.

같은 단어가 등장할 때마다 해당 단어의 숫자를 1씩 증가시키면 됩니다.

````python
sample_words = [
    "데이터",
    "분석",
    "파이썬",
    "데이터",
    "AI",
    "분석",
    "데이터",
]
````

````python
word_counts = {}
for word in sample_words:
    if word in word_counts:
        word_counts[word] += 1
    else:
        word_counts[word] = 1
print(word_counts)
````

````python
word_counts = {}

for word in sample_words:
    word_counts[word] = word_counts.get(word, 0) + 1
print(word_counts)
````

## 실습 10. 왜 형태소 분석이 필요한가?

한국어는 단순히 공백만 기준으로 단어를 나누기 어렵습니다.

한국어 텍스트 처리에서 형태소 분석이 필요한 이유

한국어는 어간에 조사나 어미가 붙어 문법적 기능을 나타내는 교착어이기 때문에, 단순히 공백 기준으로 단어를 나누면 여러 한계가 발생합니다.

1. 동일 의미 단어의 다변화

'데이터를', '데이터가', '데이터의', '데이터로'

사람은 모두 핵심 단어를 **데이터**로 인식하지만, 단순 공백 분리 시 모두 **서로 다른 문자열(토큰)**로 인식됩니다.

2. 불필요한 조사/어미의 포함

'은', '는', '이', '가', '을', '를', '의'

빈도 분석 시 단어의 핵심 의미를 담지 않는 조사가 상위에 집계되어 분석 품질을 떨어뜨립니다.

## 실습 11. Kiwi 형태소 분석기 설치하기

한글 형태소 분석 도구는 여러 가지가 있습니다.

## 실습 12. Kiwi로 한 문장 품사 확인하기

설치 후 먼저 짧은 문장을 테스트합니다.

````python
from kiwipiepy import Kiwi

kiwi = Kiwi()
sample_title = "AI 시대의 데이터 분석을 위한 파이썬"

# 형태소 분석 및 토큰 추출
tokens = kiwi.tokenize(sample_title)

# 형태소와 품사 태그 출력
for token in tokens:
    print(token.form, token.tag)
````

Kiwi 형태소 분석 결과

입력한 문장("AI 시대의 데이터 분석을 위한 파이썬")에 대한 형태소 및 품사 태그 상세 분석 내역입니다.

1. 형태소 단위 분석 내역
AI (SL): 외국어 / 알파벳

시대 (NNG): 일반 명사

의 (JKG): 관형격 조사

데이터 (NNG): 일반 명사

분석 (NNG): 일반 명사

을 (JKO): 목적격 조사

위하 (VV): 동사 (원형: '위하다')

ㄴ (ETM): 관형사형 전성 어미

파이썬 (NNP): 고유 명사

2. 핵심 분석 대상 키워드 추출 (NNG, NNP, SL)
추출된 주제 키워드

AI (SL) · 시대 (NNG) · 데이터 (NNG) · 분석 (NNG) · 파이썬 (NNP)

결과 평가: 조사(JKG, JKO), 동사/어미(VV, ETM)가 깨끗하게 분리되어 핵심 주제를 담은 명사 및 외래어 키워드만 정밀하게 선별되었습니다.

## 실습 13. 명사 중심으로 단어 추출하기

이제 사용할 품사를 정하고 함수를 만듭니다.

````python
TARGET_TAGS = {"NNG", "NNP", "SL"}
def extract_words(text):
    result = []
    for token in kiwi.tokenize(str(text)):
        if token.tag in TARGET_TAGS:
            word = token.form.strip().lower()
            result.append(word)
    return result


extract_words("AI 시대의 데이터 분석을 위한 파이썬")
````

함수가 하는 일

문장을 Kiwi로 분석

→ 형태소 하나씩 확인

→ NNG, NNP, SL만 선택

→ 앞뒤 공백 제거

→ 영문은 소문자로 통일

→ 리스트로 반환

## 실습 14. 최소 글자수 필터링 적용하기

한 글자 단어가 모두 의미 없는 것은 아닙니다.

하지만 도서 제목 빈도 분석에서는 한 글자 토큰이 지나치게 많이 등장하면 결과를 해석하기 어려울 수 있습니다.

````python
MIN_LENGTH = 2
def extract_words(text):
    result = []
    for token in kiwi.tokenize(str(text)):
        if token.tag in TARGET_TAGS:
            word = token.form.strip().lower()
            if len(word) >= MIN_LENGTH:
                result.append(word)
    return result
````

## 실습 15. 모든 도서 제목에서 명사 추출하기

````python
filtered_words = []

for title in titles:
    words = extract_words(title)
    filtered_words.extend(words)

print("추출된 전체 단어 수:", len(filtered_words))
print(filtered_words[:50])
````

````python
noun_counter = Counter(filtered_words)
noun_counter.most_common(30)
````

````python
noun_top30 = pd.DataFrame(
    noun_counter.most_common(30),
    columns=["단어", "빈도"]

)
noun_top30
````

## 실습 16. 불용어 이해하기

형태소 분석 후에도 분석 목적과 관계가 적은 단어가 상위에 나타날 수 있습니다.

## 실습 17. 불용어 목록 만들기

처음에는 너무 많은 단어를 넣지 않습니다.

````python
TARGET_TAGS = {"NNG", "NNP", "SL"}

MIN_LENGTH = 2

STOPWORDS = {
    "세트",
    "개정판",
    "에디션",
    "리커버",
    "한정판",
}

def extract_words(text):
    result = []
    for token in kiwi.tokenize(str(text)):
        if token.tag not in TARGET_TAGS:
            continue
        word = token.form.strip().lower()
        if len(word) < MIN_LENGTH:
            continue
        if word in STOPWORDS:
            continue
        result.append(word)
    return result
````

## 실습 18. 최종 단어 목록 만들기

````python
final_words = []
for title in titles:
    final_words.extend(extract_words(title))

print("최종 단어 수:", len(final_words))
print(final_words[:50])


final_counter = Counter(final_words)
final_counter.most_common(30)

top30 = pd.DataFrame(
    final_counter.most_common(30),
    columns=["단어", "빈도"]
)
top30
````

````python
top30.to_csv(
    "chapter02_top30_words.csv",
    index=False,
    encoding="utf-8-sig"
)
````
