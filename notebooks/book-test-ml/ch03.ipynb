## 실습 3. 머신러닝은 왜 텍스트를 숫자로 바꿀까?

머신러닝 알고리즘은 일반적으로 숫자를 입력받아 계산합니다.

예를 들어 다음과 같은 숫자는 계산할 수 있습니다.


도서 제목

        ↓

사용된 단어 확인

        ↓

단어마다 열(column) 생성

        ↓

등장 여부 또는 등장 횟수를 숫자로 기록

        ↓

숫자 벡터

이 과정을 텍스트 벡터화(Vectorization)라고 합니다.

## 실습 4. 아주 작은 예제로 먼저 이해하기

실제 1,000권 정도의 도서 제목을 바로 사용하면 행과 열이 많아져 원리를 보기 어렵습니다.

````python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

# 1. 예제 텍스트 데이터 정의
sample_docs = [
    "파이썬 데이터 분석",
    "파이썬 머신러닝",
    "데이터 분석 입문",
]

# --------------------------------------------------
# [방법 1] CountVectorizer: 단어의 등장 빈도수(Count) 기준
# --------------------------------------------------
cv = CountVectorizer()
cv_matrix = cv.fit_transform(sample_docs)

# 결과를 보기 쉽게 DataFrame으로 변환
df_cv = pd.DataFrame(
    cv_matrix.toarray(), 
    columns=cv.get_feature_names_out(),
    index=[f"문장 {i+1}" for i in range(len(sample_docs))]
)

print("=== 1. CountVectorizer 결과 (단어 빈도수) ===")
print(df_cv)
print("\n" + "="*50 + "\n")

# --------------------------------------------------
# [방법 2] TfidfVectorizer: 단어의 중요도(TF-IDF) 기준
# --------------------------------------------------
tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(sample_docs)

# 결과를 보기 쉽게 DataFrame으로 변환
df_tfidf = pd.DataFrame(
    tfidf_matrix.toarray().round(3),  # 소수점 3자리까지 반올림
    columns=tfidf.get_feature_names_out(),
    index=[f"문장 {i+1}" for i in range(len(sample_docs))]
)

print("=== 2. TfidfVectorizer 결과 (단어 중요도) ===")
print(df_tfidf)
````

## 실습 5. Bag of Words 이해하기

앞의 방식은 단어를 하나의 주머니에 넣고 몇 번 등장했는지 세는 것과 비슷합니다.

CountVectorizer (단어 빈도수)

원리: 단어가 몇 번 나왔는지만 카운트합니다.

TfidfVectorizer (단어 중요도 가중치)

원리: 단어 빈도(TF)에 역문서 빈도(IDF)를 곱해 "특정 문장에서만 유독 자주 쓰인 핵심 키워드"에 높은 점수를 줍니다.

장점: 문서의 주제나 특징을 대표하는 단어를 훨씬 잘 찾아냅니다.

## 실습 7. 작은 예제에 CountVectorizer 적용하기

````python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer

# 1. 예제 데이터 정의
sample_docs = [
    "파이썬 데이터 분석",
    "파이썬 머신러닝",
    "데이터 분석 입문",
]

# 2. CountVectorizer 객체 생성 및 학습/변환 (실습 7번)
count_vectorizer = CountVectorizer()
X_count_sample = count_vectorizer.fit_transform(sample_docs)

# 3. 결과 확인 (DataFrame으로 변환하여 보기 쉽게 출력)
df_count_sample = pd.DataFrame(
    X_count_sample.toarray(),
    columns=count_vectorizer.get_feature_names_out(),
    index=[f"문장 {i+1}" for i in range(len(sample_docs))]
)

print("=== CountVectorizer 변환 결과 ===")
print(df_count_sample)
````

## 실습 15. 실제 데이터의 첫 번째 제목 벡터 확인하기

````python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer

# 1. 실제 데이터에서 상품명(도서 제목) 추출 및 Vectorizer 학습
titles = df_books["상품명"]

count_vectorizer = CountVectorizer()
X_count = count_vectorizer.fit_transform(titles)
count_terms = count_vectorizer.get_feature_names_out()

# --------------------------------------------------
# [실습 15] 첫 번째 제목 벡터 및 추출된 단어 확인
# --------------------------------------------------

# 1) 원본 첫 번째 도서 제목 확인
print("=== 1. 원본 첫 번째 도서 제목 ===")
print(titles.iloc[0])
print("\n" + "="*50 + "\n")

# 2) 희소 행렬의 첫 번째 행(0번)에서 0보다 큰(실제 등장한) 단어만 추출
first_row = X_count.getrow(0)
indices = first_row.indices  # 단어가 위치한 컬럼 인덱스 번호들
values = first_row.data      # 해당 단어의 등장 횟수(값)들

first_title_terms = [
    (count_terms[index], value)
    for index, value in zip(indices, values)
]

print("=== 2. Vectorizer가 추출한 단어와 빈도수 ===")
print(first_title_terms)
````

- **원본 제목:** `"소년이 온다"`
- **`getrow(0).indices` / `.data`로 추출한 단어·빈도:** `[('소년이', 1), ('온다', 1)]`

## 실습 16. 전체 데이터에서 많이 등장한 단어 확인하기

````python
import numpy as np
import pandas as pd

# 실습 16. 전체 데이터에서 많이 등장한 단어 확인하기

# 1. Count 행렬의 각 열(단어)별 합계 계산 (axis=0)
count_sums = np.asarray(X_count.sum(axis=0)).ravel()

# 2. 단어와 전체 등장 횟수를 데이터프레임으로 제작
count_summary = pd.DataFrame({
    "단어": count_terms,
    "전체등장횟수": count_sums,
})

# 3. 등장 횟수 기준 내림차순 정렬 후 인덱스 재설정
count_summary = count_summary.sort_values(
    "전체등장횟수",
    ascending=False,
).reset_index(drop=True)

# 4. 상위 30개 단어 출력
count_summary.head(30)
````

## 실습 18. Count 방식의 한계 생각해 보기

CountVectorizer는 이해하기 쉽지만 한 가지 중요한 문제가 있습니다.

실습 18 핵심 개념 요약
CountVectorizer의 한계

모든 문서에 흔하게 등장하는 단어(예: 책, 도서, 이야기, 위한 등)도 등장 횟수가 많다는 이유만으로 높은 가중치(숫자)를 받게 됩니다.

이러한 보편적인 단어들은 정작 "이 문서가 어떤 특이한 내용을 다루고 있는가?"를 구별하는 데 아무런 도움을 주지 못합니다.

TF-IDF의 출발점

TF (Term Frequency, 단어 빈도): 특정 문서 안에서 해당 단어가 얼마나 많이 등장했는가?

IDF (Inverse Document Frequency, 역문서 빈도): 모든 문서 전체를 통틀어 해당 단어가 얼마나 희귀하게 등장했는가?

결론: 단순히 빈도수만 세는 것이 아니라, "흔한 단어의 점수는 깎고, 특정 문서에만 등장하는 유니크한 단어의 점수는 높이자"는 목적에서 TF-IDF가 탄생했습니다.

## 실습 19. TF 이해하기

습 19 핵심 개념 요약
TF (Term Frequency, 단어 빈도)

개념: 하나의 특정 문서(문장) 안에서 특정 단어가 얼마나 자주 등장하는지를 나타내는 값입니다.

예시: "파이썬 파이썬 데이터 분석" 문장에서 '파이썬'의 TF 값은 '데이터'나 '분석'보다 높습니다.

⚠️ scikit-learn TF-IDF 사용 시 필수 주의사항

scikit-learn의 TfidfVectorizer는 단순 빈도 계산에 더해 L2 정규화(Normalization) 및 IDF(역문서 빈도)를 곱해서 최종 수치를 만듭니다.

따라서 최종 출력되는 TF-IDF 수치(예: 0.587, 0.321)를 단순한 단어 등장 횟수로 직접 해석해서는 안 됩니다.

## 실습 20. DF 이해하기

 실습 20 핵심 개념 요약DF (Document Frequency, 문서 빈도)개념: 전체 문서 모음 중 특정 단어가 들어있는 문서의 개수를 의미합니다.예시 (전체 문서 100개 기준):'데이터': 80개 문서에 등장 $\rightarrow$ DF가 높음 (널리 쓰이는 흔한 단어)'약동학': 2개 문서에만 등장 $\rightarrow$ DF가 낮음 (특정 분야/주제에서만 쓰이는 희귀한 단어)DF의 역할과 의미DF는 특정 단어가 전체 문서 집합에 얼마나 널리 퍼져 있는지 알려주는 지표입니다.DF가 너무 높은 단어는 일반적인 단어일 확률이 높아 문서 분류나 키워드 추출 시 중요도가 낮게 평가되는 기준이 됩니다.

## 실습 21. IDF 이해하기

# 📚 Chapter 03. 텍스트를 숫자로 바꾸는 CountVectorizer와 TF-IDF

---

## 💡 TF-IDF 핵심 개념 요약

### 1. TF (Term Frequency, 단어 빈도)
* **개념:** 특정 문서 안에서 해당 단어가 얼마나 자주 등장하는지를 나타냅니다.
* **특징:** 하나의 문서 내부에서 단어의 중요도를 평가합니다.

### 2. DF (Document Frequency, 문서 빈도)
* **개념:** 전체 문서 모음 중 특정 단어가 포함된 문서의 개수를 의미합니다.
* **특징:** 단어가 전체 문서 집합에 얼마나 널리 퍼져 있는지를 보여줍니다.

### 3. IDF (Inverse Document Frequency, 역문서 빈도)
* **개념:** DF의 역수 개념으로, 단어의 희소성을 평가하는 가중치입니다.
* **동작 원리:**
  * **흔한 단어 (DF ↑):** 문서를 구분하는 능력이 떨어지므로 **IDF 점수 감점 (↓)**
  * **희귀한 단어 (DF ↓):** 특정 문서를 구분하는 핵심 키워드이므로 **IDF 점수 가산 (↑)**

---

## 📌 TF-IDF 계산식 및 역할

$$\text{TF-IDF} = \text{TF (문서 내 단어 빈도)} \times \text{IDF (전체 희소성 가중치)}$$

* **TF-IDF의 목적:** 단순히 빈도수만 세는 `CountVectorizer`의 한계를 극복하여, 흔한 단어의 영향력은 줄이고 **문서의 특징을 잘 나타내는 유니크한 키워드**에 더 높은 점수를 부여합니다.
* **주의사항:** `scikit-learn`은 L2 정규화 및 스무딩(Smoothing)이 적용되므로, 최종 출력값을 단순 등장 횟수로 직접 해석하지 않습니다.

## 실습 22. TF-IDF를 한 문장으로 정리하기

# 💡 실습 22. TF-IDF 핵심 요약 정리

---

### 📌 TF-IDF 한 문장 정의
> **"이 문서 내에서는 자주 등장하지만, 전체 문서 모음에서는 흔하지 않은 단어에 높은 가중치를 부여하는 지표"**

---

### 🔍 두 가지 핵심 관점의 결합

1. **TF (Term Frequency)**  
   * **질문:** *이 문서 안에서 얼마나 자주 나타나는가?*
   * **역할:** 해당 문서 내부에서의 단어 비중을 반영합니다.

2. **IDF (Inverse Document Frequency)**  
   * **질문:** *전체 문서 집합에서는 얼마나 흔하거나 드문가?*
   * **역할:** 모든 문서에 다 나오는 흔한 단어의 점수를 감점(가중치 축소)시킵니다.

---

### ⚠️ 해석 시 필수 주의사항

* **상대적 중요도:** TF-IDF 점수가 높다고 해서 현실 세계에서 무조건 절대적으로 중요한 단어라는 의미는 아닙니다.
* **조건부 특징:** **"현재 수집된 문서 집합(Corpus)"**과 **"설정한 Vectorizer 조건"** 내에서 상대적으로 차별성을 가지는 단어라는 뜻입니다.

## 실습 23. 작은 예제로 TfidfVectorizer 적용하기

````python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

# 1. 예제 데이터 및 기존 CountVectorizer 결과
sample_docs = [
    "파이썬 데이터 분석",
    "파이썬 머신러닝",
    "데이터 분석 입문",
]

# (비교용) CountVectorizer 적용
count_vectorizer = CountVectorizer()
X_count_sample = count_vectorizer.fit_transform(sample_docs)
sample_count_df = pd.DataFrame(
    X_count_sample.toarray(),
    columns=count_vectorizer.get_feature_names_out(),
    index=sample_docs,
)

# 2. [실습 23] TfidfVectorizer 적용
tfidf_sample_vectorizer = TfidfVectorizer()
X_tfidf_sample = tfidf_sample_vectorizer.fit_transform(sample_docs)
tfidf_sample_terms = tfidf_sample_vectorizer.get_feature_names_out()

tfidf_sample_df = pd.DataFrame(
    X_tfidf_sample.toarray(),
    columns=tfidf_sample_terms,
    index=sample_docs,
)

# 3. 결과 출력 및 비교
print("=== 1. CountVectorizer 결과 (정수형 빈도수) ===")
print(sample_count_df)

print("\n=== 2. TfidfVectorizer 결과 (소수점 3자리 반올림) ===")
print(tfidf_sample_df.round(3))
````

## 📊 CountVectorizer vs TfidfVectorizer 결과 비교 분석

---

### 1. 주요 차이점 핵심 요약

* **값의 형태:** `Count`는 정수(등장 횟수)로 표현되는 반면, `TF-IDF`는 가중치 연산 및 L2 정규화가 적용되어 **소수**로 표현됩니다.
* **가중치 차등 부여:** `Count`에서는 모든 단어가 똑같이 `1` 점수를 받지만, `TF-IDF`에서는 **문서를 구분하는 독특한 단어일수록 더 높은 점수**를 받습니다.

### 2. 세부 분석: 왜 숫자가 다를까?

* **Count:** `'머신러닝'`과 `'파이썬'` 모두 동일하게 `1`로 평가됩니다.
* **TF-IDF:**
  * **`머신러닝` (0.796):** 전체 3개 문서 중 단 1곳에만 나오는 **희귀한 단어**이므로 가중치가 매우 높아집니다.
  * **`파이썬` (0.605):** 다른 문서에도 등장하는 흔한 단어이므로 희소성이 떨어져 상대적으로 낮게 평가됩니다.

### ② "데이터 분석 입문" 문장 분석
* **Count:** `'데이터'`, `'분석'`, `'입문'` 세 단어 모두 `1`로 동일합니다.

   * **단어 2개 문장 (`소년이 온다`):** L2 정규화 공식($\sqrt{x_1^2 + x_2^2} = 1$)에 따라 각 단어가 기본적으로 $\frac{1}{\sqrt{2}} \approx 0.707$ 내외의 가중치를 갖습니다[cite: 5].
   * **단어 3개 문장 (`말투만 바꿨을 뿐인데`):** 포함된 단어가 3개로 늘어나면서 각각 $\frac{1}{\sqrt{3}} \approx 0.577$ 수준으로 가중치가 나누어집니다[cite: 5].

2. **단어 간 가중치 차이 (`작별하지 않는다`)**
   * `'작별하지'`(0.752)가 `'않는다'`(0.659)보다 높은 값을 가집니다[cite: 5].
   * 이는 전체 199개 도서 제목 집합 내에서 **`작별하지`라는 단어가 `않는다`라는 단어보다 상대적으로 덜 등장(희소성이 높음)**하여 더 높은 IDF 가중치를 받았기 때문입니다[cite: 5]

### 📌 데이터 누수(Data Leakage) 발생 메커니즘

| 구분 | 올바른 접근법 (Train/Test 분리 후 fit) | 잘못된 접근법 (전체 데이터에 fit) |
| :---: | :--- | :--- |
| **단어 사전 구축** | Train 데이터에 포함된 단어만 수집하여 사전을 형성함 | Test 데이터에만 등장하는 단어 정보까지 사전에 미리 포함됨 |
| **IDF 값 계산** | Train 데이터 내의 문서 빈도(DF)만을 기준으로 가중치 산출 | Test 데이터의 문서 빈도까지 포함되어 IDF 수치가 왜곡됨 |
| **모델 평가** | 미지의 실제 데이터에 대한 일반화 성능을 정확히 평가 가능 | 모델 성능이 실제보다 과도하게 좋게 나오는 **낙관적 편향** 발생 |
